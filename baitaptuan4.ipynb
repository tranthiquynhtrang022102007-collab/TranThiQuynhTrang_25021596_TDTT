```python
#4a6
def is_prime(n):
    if n <= 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0 :
            return False
    return True
while True:
    try :
        a,b = map(int,input().split())
        s=0
        for x in range(a,b+1):
            if is_prime(x):
                s +=x
        print(s)
    except EOFError:
        break
```

```python
#4a10
def ham(x):
    l =1
    while x != 1 :
        if x % 2 ==0 :
            x //= 2
        else:
            x = 3 * x + 1
        l += 1
    return l
n = int(input())
max_len = 0
best_x = 0
for i in range(1, n+1):
    L = ham(i)
    if L > max_len :
        max_len= L
        best_x = i
print(best_x, max_len)
```

```python
#4a13
def sum_of_divisors(n):
    total =0
    for i in range(1,n):
        if n % i == 0 :
            total += i
    return total
a = int(input())
b = int(input())
sum_a = sum_of_divisors(a)
sum_b = sum_of_divisors(b)
if sum_a == b and sum_b == a:
    print("true")
else:
    print("false")
```
