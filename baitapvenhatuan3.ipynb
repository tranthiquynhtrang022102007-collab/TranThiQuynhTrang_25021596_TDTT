```python
#b14:
print(" nhap cac so nguyen , nhap -1 de dung:")
first = True 
while True :
    x = int(input())
    if x == -1:
        break
    if first:
        max_val = x
        min_val = x
        first = False
    else :
        if x > max_val:
            max_val =x
        if x < min_val:
            min_val = x
if first:
    print(" khong co so nao hop le ")
else:
    print("gia tri lon nhat la:" , max_val)
    print("gia tri nho nhat la:", min_val)
```
